```python
import numpy as np
harga = np.array([5000, 7000, 3000])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Kesimpulan: NumPy array dapat langsung melakukan operasi matematika pada setiap elemen, sedangkan list Python biasa tidak bisa dikalikan dengan angka desimal seperti 0.9.

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Kolom yang memiliki data kosong adalah 'menu' di index 4, dan kolom 'terjual' di index 2. Jika langsung dianalisis, hasil perhitungan seperti total penjualan atau rata-rata bisa menjadi tidak akurat karena ada data yang kosong.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total berarti kolom tersebut memiliki data kosong missing value.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kolom terjual diisi dengan 0 karena data kosong dapat dianggap tidak ada barang yang terjual.

Sedangkan baris dengan menu kosong dihapus karena nama menu merupakan data penting. Jika kosong, data tersebut sulit digunakan dalam analisis.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah drop_duplicates(), baris yang terduplikat akan di hapus. Jika tidak ada duplikat, jumlah baris tetap sama.

Memastikan tipe data (dtypes) misalnya harga harus bertipe integer agar bisa digunakan dalam operasi matematika.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total pendapatan tertinggi adalah Nasi Goreng, karena memiliki harga dan jumlah penjualan yang tinggi.

Informasi ini membantu kantin untuk menambah stok Nasi Goreng dan lebih memprioritaskannya karena menghasilkan pendapatan yang besar.
